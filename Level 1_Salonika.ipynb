Task 3: Geospatial Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\Salonika Sahoo\Downloads\Dataset .csv")

plt.figure(figsize=(8,6))
plt.scatter(df["Longitude"],df["Latitude"],alpha=0.5)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Restaurant Locations")
plt.show()
```

```python
correlation = df["Aggregate rating"].corr(df["Latitude"])
print("Latitude vs Rating:",correlation)

correlation = df["Aggregate rating"].corr(df["Longitude"])
print("Longitude vs Rating:",correlation)
```
