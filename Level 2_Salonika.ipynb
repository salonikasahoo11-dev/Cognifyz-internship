Task 1: Table Booking and Online Delivery

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Salonika Sahoo\Downloads\Dataset .csv")

table_booking = (df["Has Table booking"]=="Yes").mean()*100
online_delivery = (df["Has Online delivery"]=="Yes").mean()*100
print("Table Booking:",table_booking,"%")
print("Online Delivery:",online_delivery,"%")
```

```python
print(df.groupby("Has Table booking")["Aggregate rating"].mean())
```

```python
print(df.groupby("Price range")["Has Online delivery"].value_counts())
```

Task 2: Price Range Analysis

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Salonika Sahoo\Downloads\Dataset .csv")

print(df["Price range"].mode()[0])
```

```python
avg_rating = df.groupby("Price range")["Aggregate rating"].mean()
print(avg_rating)
```

```python
rating_color = df.groupby("Rating color")["Aggregate rating"].mean()
print(rating_color.sort_values(ascending=False))
```

Task 3: Feature Engineering

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Salonika Sahoo\Downloads\Dataset .csv")

df["Name_Length"] = df["Restaurant Name"].str.len()

df["Address_Length"] = df["Address"].str.len()
print(df[["Restaurant Name", "Name_Length", "Address", "Address_Length"]])
```

```python
df["Has_Table_Booking"] = df["Has Table booking"].map({"Yes":1,"No":0})

df["Has_Online_Delivery"] = df["Has Online delivery"].map({"Yes":1,"No":0})

print(df[["Has Table booking", "Has_Table_Booking", "Has Online delivery", "Has_Online_Delivery"]])
```
